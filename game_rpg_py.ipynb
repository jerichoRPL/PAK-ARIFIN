### Eksperimen Tanpa `super()`
Mari kita lihat apa yang terjadi jika kita memutus hubungan inisialisasi antara Anak (`Mage`) dan Induk (`Hero`).

```python
from abc import ABC, abstractmethod


# Blueprint utama: semua unit game wajib bisa menyerang dan menampilkan info.
class GameUnit(ABC):
	@abstractmethod
	def serang(self, target):
		pass

	@abstractmethod
	def info(self):
		pass


class Hero(GameUnit):
	# Constructor: Dijalankan saat Hero baru dibuat
	def __init__(self, name, hp=100, attack_power=0):
		self.name = name # Nama Hero
		self.__hp = hp # Nyawa (Health Point) bersifat private
		self.attack_power = attack_power # Kekuatan Serangan

	# GETTER: Cara resmi melihat HP
	def get_hp(self):
		return self.__hp

	# SETTER: Cara resmi mengubah HP dengan validasi
	def set_hp(self, nilai_baru):
		if nilai_baru < 0:
			self.__hp = 0
		elif nilai_baru > 1000:
			print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
			self.__hp = 1000
		else:
			self.__hp = nilai_baru

	# Method untuk menampilkan info hero
	def info(self):
		print(f"Hero: {self.name} | HP: {self.get_hp()} | Power: {self.attack_power}")

	# Method menyerang: Objek ini (self) menyerang objek lain (lawan)
	def serang(self, lawan):
		print(f"{self.name} menyerang {lawan.name}!")
		lawan.diserang(self.attack_power)

	# Method diserang: Menerima damage
	def diserang(self, damage):
		sisa_hp = self.get_hp() - damage
		self.set_hp(sisa_hp)
		print(f"{self.name} terkena damage {damage}. Sisa HP: {self.get_hp()}")


# Class Mage adalah anak dari class Hero
class Mage(Hero):
	def __init__(self, name, hp=100, attack_power=0, mana=100):
		# Memanggil constructor milik Parent (Hero)
		super().__init__(name, hp, attack_power)
		self.mana = mana

	def info(self):
		print(f"{self.name} [Mage] | HP: {self.get_hp()} | Mana: {self.mana}")

	def serang(self, lawan=None):
		if lawan is None:
			print(f"{self.name} (Mage) menembakkan Bola Api! Boom!")
			return
		super().serang(lawan)

	# Mage punya skill khusus
	def skill_fireball(self, lawan):
		if self.mana >= 20:
			print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
			self.mana -= 20
			lawan.diserang(self.attack_power * 2) # Damage 2x lipat
		else:
			print(f"{self.name} gagal skill! Mana tidak cukup.")


class Monster(GameUnit):
	def __init__(self, jenis):
		self.jenis = jenis

	def serang(self, target):
		print(f"Monster {self.jenis} menggigit {target}!")

	def info(self):
		print(f"Saya adalah Monster: {self.jenis}")


class Archer(Hero):
	def serang(self, lawan=None):
		if lawan is None:
			print(f"{self.name} (Archer) memanah dari jauh! Jleb!")
		else:
			super().serang(lawan)


class Fighter(Hero):
	def serang(self, lawan=None):
		if lawan is None:
			print(f"{self.name} (Fighter) memukul dengan pedang! Slash!")
		else:
			super().serang(lawan)


# -- Main Program --
# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)
# Memanggil Method
hero1.info()
hero2.info()

# Tambah kode Output di akhir program
print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2) # Layla menyerang Zilong
hero2.serang(hero1) # Zilong membalas

# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)
eudora.info()
eudora.serang(balmond) # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage

# -- Uji Enkapsulasi HP --
print("\n--- Uji Enkapsulasi HP ---")
hero_encapsulation = Hero("Layla", 100)
hero_encapsulation.set_hp(-50)
print(f"HP setelah diatur ke -50: {hero_encapsulation.get_hp()}")
hero_encapsulation.set_hp(9999)
print(f"HP setelah diatur ke 9999: {hero_encapsulation.get_hp()}")

# -- Uji Abstraction & Interface --
print("\n--- Uji Abstraction & Interface ---")
# GameUnit() # Error: abstract class tidak bisa dibuat menjadi objek.
abstract_hero = Hero("Alucard")
monster = Monster("Serigala")
abstract_hero.info()
monster.info()
monster.serang("Alucard")

# -- Uji Polymorphism --
pasukan = [
	Mage("Eudora"),
	Archer("Miya"),
	Fighter("Zilong"),
	Mage("Gord")
]

print("\n--- PERANG DIMULAI ---")
for pahlawan in pasukan:
	pahlawan.serang()
```

### Jawaban Pertanyaan:

*   **Error apa yang muncul?** Muncul `AttributeError: 'MageTanpaSuper' object has no attribute 'nama'`.
*   **Mengapa error terjadi padahal kita mengirim nama "Eudora"?** Karena argumen `"Eudora"` hanya masuk ke parameter fungsi `__init__` milik `MageTanpaSuper`, tapi tidak pernah diproses lebih lanjut. Pada Python, jika class Anak mendefinisikan `__init__`, ia akan **menimpa** (override) `__init__` milik Induk sepenuhnya. Tanpa `super()`, baris `self.nama = nama` yang ada di class `Hero` tidak pernah dieksekusi.

**Peran `super()`:**
Fungsi `super()` berfungsi untuk memanggil method dari class induk (Parent). Dalam konstruktor (`__init__`), `super()` memastikan bahwa data yang dikirim dari objek Anak diteruskan ke logika Induk. Ini memungkinkan kita untuk **menggunakan kembali (reuse)** kode yang sudah ada di class Induk sehingga kita tidak perlu menulis ulang kode inisialisasi dasar berkali-kali di setiap class turunan.
